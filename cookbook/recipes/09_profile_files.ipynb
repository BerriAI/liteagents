## 3. Write a YAML profile

Edit the values below to configure your agent. The file contains no API key; the SDK reads the key you entered earlier.

```python
import tempfile
from pathlib import Path

workspace = Path(tempfile.mkdtemp(prefix="liteagents-"))

yaml_path = workspace / "agent.yaml"
yaml_path.write_text("""harness: pydantic-ai
model: openai/gpt-5.4-mini
system_prompt: Be concise.
""")
print(yaml_path.read_text())
```

## 4. Load it and run

`from_yaml()` reads the file and validates the options. The `run()` call is unchanged.

```python
from liteagents import ProfileOptions, run

profile = ProfileOptions.from_yaml(yaml_path)
prompt = "Explain what an agent harness does in one sentence."
result = await run(prompt, profile=profile)
print(result.text)
```

## 5. Try the same profile as JSON

Use `from_json()` for a JSON file. It creates the same profile as the YAML above.

```python
json_path = workspace / "agent.json"
json_path.write_text('''{
  "harness": "pydantic-ai",
  "model": "openai/gpt-5.4-mini",
  "system_prompt": "Be concise."
}''')
json_profile = ProfileOptions.from_json(json_path)
print("Same profile:", profile == json_profile)
result = await run(prompt, profile=json_profile)
print(result.text)
```

Try changing `system_prompt` in either file. Rerun its write and load cells to apply the change.
