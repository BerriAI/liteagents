## 3. Enable streaming

Use `LiteAgentClient` for conversation history and `TextDelta` for text as it arrives.

```python
from liteagents import LiteAgentClient, ProfileOptions, TextDelta

profile = ProfileOptions(
    harness="pydantic-ai",
    model="openai/gpt-5.4-mini",
)
profile.features.streaming = True
```

## 4. Ask two questions

Both calls use the same client, so the second answer should remember **COBALT-42** from the first turn.

```python
async with LiteAgentClient(profile=profile) as agent:
    print("First answer:")
    async for message in agent.query("Remember my project code: COBALT-42. Confirm you have it."):
        if isinstance(message, TextDelta):
            print(message.text, end="", flush=True)

    print("\n\nFollow-up:")
    async for message in agent.query("What project code did I give you?"):
        if isinstance(message, TextDelta):
            print(message.text, end="", flush=True)
    print()
```

Change the code or the follow-up question and rerun the last cell.

Keep one client open for a conversation. Use `run()` for independent tasks where you only need the final answer.
