## Create your profile

This is the SDK interface. The following cells change this profile to demonstrate
one feature. A temporary workspace keeps each run's files separate.

```python
import asyncio
import tempfile
from pathlib import Path

from liteagents import LiteAgentClient, LiteAgentOptions, ProfileOptions

profile = ProfileOptions(
    harness=HARNESS,
    model=MODEL,
    model_kwargs=MODEL_KWARGS,
    tools=[],
    max_turns=10,
    system_prompt="Use the requested tools and report their actual results. Be concise.",
)
workspace_root = Path(os.environ.get(
    "LITEAGENTS_NOTEBOOK_WORKSPACE", Path(tempfile.gettempdir()) / "liteagents-notebooks"
))
workspace_root.mkdir(parents=True, exist_ok=True)
workspace = Path(tempfile.mkdtemp(prefix="run-", dir=workspace_root)).resolve()
print("Workspace:", workspace)
```

```python
ORDER_ID = "A123"
```

## Define the tool

Edit `execute()` to connect your own data source. This fixture returns demo data and records calls
in memory so you can inspect them after the run.

```python
from typing import ClassVar

from liteagents import AssistantMessage, Tool, ToolUseBlock


class LookupOrder(Tool):
    name = "lookup_order"
    description = "Return the payment status of an order."
    input_schema: ClassVar[dict] = {
        "type": "object",
        "properties": {"order_id": {"type": "string"}},
        "required": ["order_id"],
        "additionalProperties": False,
    }

    def __init__(self):
        self.calls = []

    async def execute(self, input):
        self.calls.append(input)
        return f"Order {input['order_id']}: paid, total USD 12"
```

```python
profile.tools = ["lookup_order"]
tool = LookupOrder()
async with asyncio.timeout(180):
    async with LiteAgentClient(
        options=LiteAgentOptions(profile=profile, cwd=workspace, tools=[tool])
    ) as client:
        run = await client.start_run(
            f"Call lookup_order for {ORDER_ID} and report its payment status and total."
        )
        result = await run.result()
for message in result.messages:
    if isinstance(message, AssistantMessage):
        for block in message.content:
            if isinstance(block, ToolUseBlock):
                print("Tool:", block.name, block.input)
                assert block.name == "lookup_order"
print(result.text)
assert {"order_id": ORDER_ID} in tool.calls
```
