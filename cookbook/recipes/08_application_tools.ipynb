## 3. Define the order lookup

Write a Python function with type hints and a docstring. LiteAgents generates its tool schema and adapts it to the selected harness. Both regular and `async` functions work.

```python
def lookup_order(order_id: str) -> str:
    """Look up an order's payment status and total."""
    print(f"Looking up {order_id}")
    return f"Order {order_id}: paid, total USD 12"
```

## 4. Give the tool to your agent

Pass the tool to `run()`. You should see **Looking up A123**, followed by an answer using the payment status and total.

```python
from liteagents import ProfileOptions, run

profile = ProfileOptions(
    harness="pydantic-ai",
    model="openai/gpt-5.4-mini",
)

result = await run(
    "Look up order A123 and tell the customer its payment status and total.",
    profile=profile,
    tools=[lookup_order],
)
print(result.text)
```
